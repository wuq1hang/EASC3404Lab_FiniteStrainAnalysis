## Spatial distribution of deformation: homogeneous case
In this experment, instead of focusing on one strain ellipsoid, we compute the deformation for a hypothetical outcrop with the dimension of $[-1.0, 1.0] \times [-1.0, 1.0]$ cm subject to a spatially homogeneous simple shear deformation with $\gamma=\tan\theta$ and verify whether this deformation field can explain the structural pattern we see in a natural shear zone.

```python
# Import the necessary libraries:
import numpy as np # linear algebra and numerical operations
import matplotlib.pyplot as plt # plotting library
```

```python
def generate_circle(center, radius):
    """
    Generates the x and y coordinates of a circle given a center and radius.
    """
    xy = [] # Python built-in list to hold the x and y coordinates of the circle

    for i in range(0, 360, 1):
        angle = np.radians(i)
        x = center[0] + radius * np.cos(angle)
        y = center[1] + radius * np.sin(angle)
        xy.append([x, y])

    return np.array(xy) # convert to np.array data type for easier manipulation
```

```python
def apply_deformation(points, gamma):
    """
    Apply a homogeneous deformation to a set of points.
    """
    deformed_points = []

    ############################################
    # use the deformation matrix F for simple shear
    F = np.array([[1.0, 0.0], [0.0, 1.0]])
    ############################################

    for xy in points:
        # apply deformation to each point
        # and store the results in deformed_points variable
        deformed_points.append(F @ xy)

    return np.array(deformed_points)
```

```python

gamma = np.tan(np.radians(45))# shear strain value
x = np.linspace(-1.0, 1.0, 10)
y = np.linspace(-1.0, 1.0, 10)
X,Y = np.meshgrid(x,y) 
fig, ax = plt.subplots()

for (x,y) in zip(X.flatten(), Y.flatten()):
    center = np.array([x,y])
    xy_circle = generate_circle(center, 0.1)

    xy_deformed = apply_deformation(xy_circle, gamma)
    ax.plot(xy_deformed[:,0], xy_deformed[:,1], color='blue')

ax.set_aspect('equal')
```
